```python
import cfbd
import pandas as pd

#  API key
configuration = cfbd.Configuration(
    access_token = '8rmahp8SVlwiZm3kLJXO03EMbAHh5N0AXFVZTEWoLV5XXImjeVk73YdiplfEM++E'
)

# Initialize the API client
api_client = cfbd.ApiClient(configuration)

# 2. Tell the API we specifically want to look at "Plays" today
plays_api = cfbd.PlaysApi(api_client)

# 3. Ask for all the plays from a specific team's season (e.g., 2023 Michigan)
print("Asking the server for the plays... this might take a few seconds...")
raw_plays = plays_api.get_plays(year=2023, season_type='regular', team='Michigan', week=12)

# 4. Extract just the columns we care about to keep our table clean
play_list = []
for play in raw_plays:
    play_list.append({
        'Quarter': play.period,
        'Clock': f'{play.clock.minutes}:{play.clock.seconds:02d}' if play.clock else 'N/A',
        'Down': play.down,
        'Distance': play.distance,
        'Play Description': play.play_text
    })

# 5. Turn that list into a pandas DataFrame (our "Excel on steroids" table)
df_plays = pd.DataFrame(play_list)

# 6. Display the first 10 rows of our beautiful new table!
display(df_plays.head(10))
```
